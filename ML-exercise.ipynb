# Functions You'll Need to Use

## pandas: Loading and Looking
pd.read_csv() - loads a csv file into a DataFrame (table)

.head() - returns the first 5 rows

.shape - returns the dimensions of the table 
(# rows, # columns)

.describe() - shows stats per column such as max, min, average, etc.

.dtypes - returns each column's data type

len() - returns number of rows

## pandas: Cleaning Things Up

pd.to_datetime() - changes data type of text into a real date

.duplicated().sum() - returns number of duplicate rows

.drop_duplicates() - removes duplicate rows

.isna().sum() - returns number of missing values per column

.dropna() - removes rows with any missing value

np.nan - "missing value"

.loc[rows, columns] - selects and chnages specific rows and columns



```python
# import statements
import pandas as pd # for data
import numpy as np # for math
import sklearn as learn # for machine learning
import matplotlib.pyplot as plt # for graphs
```

```python
df = pd.read_csv("bike_rentals.csv") # reads CSV file into a pandas dataframe
df.head() # shows first 5 rows
```

```python
df["hour"] = pd.to_datetime(df["timestamp"]).dt.hour # turns timestamp into a datetime object and extracts the hour
df["day_of_week"] = pd.to_datetime(df["timestamp"]).dt.dayofweek
```
